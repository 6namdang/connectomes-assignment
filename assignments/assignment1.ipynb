```python
import numpy as np
from cloudvolume import CloudVolume
```

```python
cv = CloudVolume(
   'precomputed://https://bossdb-open-data.s3.amazonaws.com/iarpa_microns/minnie/minnie65/em', 
    progress=True, # shows progress bar
    use_https=True,
    cache=True, # cache to disk to avoid repeated downloads
    # parallel=True, # uncomment to try parallel download!
)
```

```python
print(cv.bounds)
print(cv.resolution)
```

203986, 173170, 21075

These are the coordinators that I got from the data image that you sent. These coordinators are from mip 0, so I had to convert this point to 
mip 4

```python
cv.mip = 4

mip4_points = cv.point_to_mip(
    (203986, 173170, 21075),
    mip=0,
    to_mip=4,
)
print(mip4_points)

z = mip4_points[2]
image = cv[:, :, z:z+1]
```
